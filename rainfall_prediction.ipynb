```python
day_index = 798  # change if needed

y = df["PrecipitationSumInches"]
days = range(len(y))

print("The precipitation trend graph:")

plt.scatter(days, y, color='g')                       # all points
plt.scatter(day_index, y.iloc[day_index], color='r')  # highlight one point
plt.title("Precipitation level")
plt.xlabel("Days")
plt.ylabel("Precipitation in inches")
plt.show()
```

```python
x_vis = df.filter([
    'TempAvgF',
    'DewPointAvgF',
    'HumidityAvgPercent',
    'SeaLevelPressureAvgInches',
    'VisibilityAvgMiles',
    'WindAvgMPH'
])
```

```python

print("The precipitation vs attributes trend graph:")

for i in range(x_vis.columns.size):
    plt.subplot (3,2,i+1)
    plt. scatter(days, x_vis[x_vis.columns.values[i][:100]], color = 'g')
    plt. scatter(days[day_index], x_vis[x_vis.columns.values[i]][day_index], color = 'r' )
    plt.title(x_vis.columns.values[i])
plt.show()
```
